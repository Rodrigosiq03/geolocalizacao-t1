```python
import requests, json, os

# diretórios
os.makedirs("data/bike", exist_ok=True)

# CityBikes
URL = "http://api.citybik.es/v2/networks/bikesampa"
r = requests.get(URL); r.raise_for_status()
with open("data/bike/stations_citybikes.json","w") as f:
    json.dump(r.json(), f, indent=2, ensure_ascii=False)

# GBFS discovery
GBFS = "https://saopaulo.publicbikesystem.net/customer/gbfs/v2/gbfs.json"
r = requests.get(GBFS); r.raise_for_status()
feeds = {f['name']:f['url'] for f in r.json()['data']['en']['feeds']}

# station_information
r2 = requests.get(feeds['station_information']); r2.raise_for_status()
with open("data/bike/station_information.json","w") as f:
    json.dump(r2.json(), f, indent=2, ensure_ascii=False)

# station_status
r3 = requests.get(feeds['station_status']); r3.raise_for_status()
with open("data/bike/station_status.json","w") as f:
    json.dump(r3.json(), f, indent=2, ensure_ascii=False)

print("Arquivos em data/bike:", os.listdir("data/bike"))
```
